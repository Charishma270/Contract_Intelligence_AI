```python
all_labels = set()

for contract in contracts:
    for paragraph in contract["paragraphs"]:
        for qa in paragraph["qas"]:
            
            label = qa["id"].split("__")[-1]
            all_labels.add(label)

print(len(all_labels))
print(all_labels)
```

```python
from collections import Counter

label_counter = Counter()

for contract in contracts:
    for paragraph in contract["paragraphs"]:
        for qa in paragraph["qas"]:
            
            label = qa["id"].split("__")[-1]
            label_counter[label] += 1

print(label_counter.most_common(10))
```

```python
impossible_count = 0
possible_count = 0

for contract in contracts:
    for paragraph in contract["paragraphs"]:
        for qa in paragraph["qas"]:
            
            if qa["is_impossible"]:
                impossible_count += 1
            else:
                possible_count += 1

print("Possible:", possible_count)
print("Impossible:", impossible_count)
```

```python
target_labels = [
    "Termination For Convenience",
    "Renewal Term",
    "Cap On Liability",
    "Uncapped Liability"
]

for label in target_labels:
    print(label, ":", label_counter[label])
```

```python
target_label = "Termination For Convenience"

for contract in contracts:
    for paragraph in contract["paragraphs"]:
        for qa in paragraph["qas"]:

            label = qa["id"].split("__")[-1]

            if label == target_label and not qa["is_impossible"]:

                print("QUESTION:")
                print(qa["question"])

                print("\nANSWER:")
                print(qa["answers"][0]["text"])

                print("\nCONTEXT:")
                print(paragraph["context"][:2000])

                break
```
